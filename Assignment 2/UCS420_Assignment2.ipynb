## Q1 — Lists

```python
roll = 1024170446
digits = [int(d) for d in str(roll)]
L = [d * 10 for d in digits]
print("L:", L)

L.append(55)
print("After append:", L)   # 55 added at the end

L.insert(3, 25)
print("After insert at index 3:", L)   # 25 inserted, rest shifted right

L.remove(0)
print("After remove(0):", L)   # first occurrence of 0 removed

popped = L.pop()
print("After pop():", L, "| popped:", popped)   # last element removed

L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

print("First three:", L[:3])
print("Last three:", L[-3:])

avg = sum(L) / len(L)
greater = [x for x in L if x > avg]
print("Greater than avg (", round(avg, 2), "):", greater)
```

## Q2 — Tuples

```python
# rebuild L up to the pop step (before sorting)
roll = 1024170446
digits = [int(d) for d in str(roll)]
L = [d * 10 for d in digits]
L.append(55)
L.insert(3, 25)
L.remove(0)
L.pop()

scores = tuple(L[:8])
print("scores:", scores)

print("Highest:", max(scores), "at index", scores.index(max(scores)))
print("Lowest:", min(scores), "appears", scores.count(min(scores)), "time(s)")

# tuples are immutable — no in-place reverse possible
rev = list(scores[::-1])
print("Reversed as list:", rev)

# scores[0] = 100   → TypeError: 'tuple' object does not support item assignment
# tuples can't be changed after creation; lists can

first, second, *rest = scores
print("first:", first, "second:", second, "rest:", rest)

# uncomment to run interactively:
val = int(input("Enter a score: "))
if val in scores:
    print("First occurrence at index", scores.index(val))
else:
    print("Not present")
```

## Q3 — Random numbers

```python
import random
from collections import Counter

random.seed(42)
nums = [random.randint(100, 900) for _ in range(100)]

odd = [n for n in nums if n % 2]
even = [n for n in nums if n % 2 == 0]
print("Odd count:", len(odd))
print("Even count:", len(even))

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [n for n in nums if is_prime(n)]
print("Prime count:", len(primes))
print("Primes:", primes)

most, times = Counter(nums).most_common(1)[0]
print("Most frequent:", most, "appears", times, "times")
```

## Q4 — Sets

```python
roll = 1024170446
digits8 = [int(d) for d in str(roll)[:8]]

A = {d * 7 for d in digits8}
B = {d * 9 for d in digits8}
print("A:", A)
print("B:", B)

print("Union:", A | B)
print("Intersection:", A & B)
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference keeps only one side; symmetric_difference keeps unique elements from both
print("Symmetric difference:", A.symmetric_difference(B))

print("A subset of B?", A.issubset(B))
print("B superset of A?", B.issuperset(A))

X = int(input("Enter X: "))
A.discard(X)   # discard never raises error if missing; remove does
print("After discard:", A)
```
