## Q6 — Employees dataset

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp = pd.DataFrame(emp_data)
emp.to_csv("employees.csv", index=False)
emp
```

### a) Shape

```python
print("Shape:", emp.shape)
```

### b) Summary (info)

```python
emp.info()
```

### c) Descriptive statistics

```python
emp.describe()
```

### d) First 5 and last 3 rows

```python
print("First 5:")
display(emp.head())
print("\nLast 3:")
display(emp.tail(3))
```

### e) Average salary, total bonus, youngest age, highest rating

```python
print("Average salary:", emp["Salary"].mean())
print("Total bonus:", emp["Bonus"].sum())
print("Youngest age:", emp["Age"].min())
print("Highest rating:", emp["Rating"].max())
```

### f) Sort by Salary descending

```python
emp.sort_values("Salary", ascending=False)
```

### g) Performance category

```python
def rating_cat(r):
    if r >= 4.5:
        return "Excellent"
    elif r >= 4.0:
        return "Good"
    else:
        return "Average"

emp["Performance"] = emp["Rating"].apply(rating_cat)
emp[["Name", "Rating", "Performance"]]
```

### h) Missing values

```python
emp.isnull().sum()
```

### i) Rename Employee_ID → ID

```python
emp = emp.rename(columns={"Employee_ID": "ID"})
emp.head()
```

### j) Experience > 5 and IT department

```python
print("Experience > 5:")
display(emp[emp["Years_of_Experience"] > 5])

print("\nIT department:")
display(emp[emp["Department"] == "IT"])
```

### k) Add Tax column (10% of Salary)

```python
emp["Tax"] = emp["Salary"] * 0.10
emp[["Name", "Salary", "Tax"]]
```

### l) Save modified DataFrame

```python
emp.to_csv("employees_modified.csv", index=False)
print("Saved to employees_modified.csv")
emp
```
