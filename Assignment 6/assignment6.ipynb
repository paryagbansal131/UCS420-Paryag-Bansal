## Q1 — Sensor readings (vectorization)

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original:", temperature)

# a. add 2°C
corrected = temperature + 2
print("After +2°C:", corrected)

# b. C to F
fahrenheit = (9/5) * temperature + 32
print("Fahrenheit:", fahrenheit)

# c. readings > 32
hot = temperature[temperature > 32]
print("Above 32°C:", hot)

# d. count
print("Count above 32°C:", np.sum(temperature > 32))

# e. vectorization & boolean indexing skip python loops —
# numpy does the work in compiled C, so large arrays stay fast
```

## Q2 — Daily steps matrix

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print("steps:\n", steps)

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Max:", np.max(steps), "Min:", np.min(steps))
print("Total per day (axis=0):", np.sum(steps, axis=0))
print("Total per user (axis=1):", np.sum(steps, axis=1))

# argmax gives flat index; unravel to (user, day)
flat_idx = np.argmax(steps)
pos = np.unravel_index(flat_idx, steps.shape)
print("Max at user/day:", pos, "value:", steps[pos])
```

## Q3 — Slicing, views, reshape, flatten vs ravel

```python
original = np.array([1, 2, 3, 4, 5, 6])
print("original:", original)

# b. slice index 1 to 4
subset = original[1:5]
print("subset:", subset)

# c. modify subset (view) — original changes too
subset[0] = 999
print("After subset[0]=999")
print("original:", original)
print("subset:", subset)

# d. copy — original stays safe
original = np.array([1, 2, 3, 4, 5, 6])  # reset
copied = original[1:5].copy()
copied[0] = 500
print("\nAfter copy modify")
print("original:", original)
print("copied:", copied)

# e. arange + reshape
mat = np.arange(1, 13).reshape(3, 4)
print("\n3x4 matrix:\n", mat)

# f. extractions
print("First row:", mat[0])
print("Last row:", mat[-1])
print("Second column:", mat[:, 1])
print("Rows 1-2, cols 2-3:\n", mat[1:3, 2:4])

# g. flatten vs ravel
flat = mat.flatten()
rav = mat.ravel()
print("\nflatten():", flat)
print("ravel():", rav)

# h. ravel is a view — original can change
rav[0] = 100
print("After ravel[0]=100, matrix:\n", mat)

# i. flatten is a copy — original stays
mat = np.arange(1, 13).reshape(3, 4)  # reset
flat = mat.flatten()
flat[0] = 200
print("After flatten[0]=200, matrix:\n", mat)

# j. attributes
print("shape:", mat.shape)
print("ndim:", mat.ndim)
print("size:", mat.size)
print("dtype:", mat.dtype)
```

## Q4 — OLS regression for assistance score

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("X shape:", X.shape, "ndim:", X.ndim)

# b. transpose — features become rows, users become columns
print("X.T:\n", X.T)

# c. X.T @ X
XtX = X.T @ X
print("X.T @ X:\n", XtX)

# d. inverse
XtX_inv = np.linalg.inv(XtX)
print("inv(X.T @ X):\n", XtX_inv)

# e. beta = (X.T X)^-1 X.T y
beta = XtX_inv @ X.T @ y
print("coefficients beta:", beta)

# f. beta[0] = weight for sleep hours
#    beta[1] = weight for activity level
#    beta[2] = weight for stress level
# positive weight → higher feature pushes score up, negative → pulls it down

# g. predict for new user
new_user = np.array([5, 40, 7])
pred = new_user @ beta
print("Predicted assistance score:", pred)
```
