```python
def sum_of_primes(n):
    prime_sum = 0
    for nb in range(2, n + 1):
        f = 0
        for i in range(2, int(nb**0.5) + 1):
            if nb % i == 0:
                f = 1
                break
        if f == 0:
            prime_sum += nb
    return prime_sum


n = int(input("Enter the value of N = "))
print("The sum of all prime numbers up to ",n," is: ",sum_of_primes(n))
```

### **15.3 Add all the numbers given at cmd line**
**Note:** To run the program at cmd line
*   python Program.py
*   python Program.py 10
*   python Program.py 10 20 30 40

---

```python
import sys
print (sys.argv)
sum=0
for s in sys.argv[1:]:
	sum += int(s)

print ("Sum is = ", sum)
```

# **16 File Handling**

**Learning:** How to open the file, read the file and write in the file

---
### **16.1 Writing 1 to 10 in file**

```python
fp=open('result.txt','w')
for  i in range(1,11):
	fp.write(str(i) + "\n")
 
fp.write("Paryag Bansal\n")
fp.close()

print ("Writing done !! \nOpen result.txt to view the content")
```

---
### **16.2 Read a file and print its content**
---

```python
fp=open('result.txt')
for line in fp:
	print (line.strip())
fp.close()
```

---
### **16.3 Read from one file, Convert it to upper case and write to other file**
---

```python
Readfp=open('result.txt')
Writefp=open('r2.txt','w')
for line in Readfp:
	Writefp.write(line.upper())

Writefp.close()
Readfp.close()

print ("Writing done !! \nOpen result.txt to view the content")
```
